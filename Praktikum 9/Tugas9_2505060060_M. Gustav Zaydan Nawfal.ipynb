B. Kueri Bertingkat (Read) 
-

```python
# Menampilkan Menu Kopi dengan Tingkat Kepahitan di Atas 3

hasil_kopi = koleksi_menu.find({
    "kategori": "Kopi",
    "tingkat_kepahitan": {
        "$gt": 3
    }
})

print("Menu Kopi dengan tingkat kepahitan di atas 3:")

for menu in hasil_kopi:
    print(
        menu["nama_menu"],
        "| Harga: Rp",
        menu["harga"],
        "| Tingkat Kepahitan:",
        menu["tingkat_kepahitan"]
    )
```

```python
# Menampilkan Menu dengan Harga Rp15.000 sampai Rp30.000

hasil_harga = koleksi_menu.find({
    "harga": {
        "$gte": 15000,
        "$lte": 30000
    }
})

print("Menu dengan harga Rp15.000 sampai Rp30.000:")

for menu in hasil_harga:
    print(
        menu["nama_menu"],
        "| Kategori:",
        menu["kategori"],
        "| Harga: Rp",
        menu["harga"]
    )
```

```python
# Menampilkan Menu yang Tidak Memiliki Field pedas_level

hasil_tanpa_pedas = koleksi_menu.find({
    "pedas_level": {
        "$exists": False
    }
})

print("Menu yang tidak memiliki field pedas_level:")

for menu in hasil_tanpa_pedas:
    print(
        menu["nama_menu"],
        "| Kategori:",
        menu["kategori"],
        "| Harga: Rp",
        menu["harga"]
    )
```

C. Pembaruan Data (Update)
-

```python
# Menambahkan Status ke Seluruh Menu

hasil_update = koleksi_menu.update_many(
    {},
    {
        "$set": {
            "status": "Tersedia"
        }
    }
)

print("Jumlah dokumen yang diperbarui:", hasil_update.modified_count)

print("\nDaftar status seluruh menu:")

for menu in koleksi_menu.find(
    {},
    {"_id": 0, "nama_menu": 1, "status": 1}
):
    print(menu)
```

```python
# Mengubah Status Dua Menu Menjadi Habis

# Mengubah status Es Kopi Nusantara
hasil_update1 = koleksi_menu.update_one(
    {"nama_menu": "Es Kopi Nusantara"},
    {
        "$set": {
            "status": "Habis"
        }
    }
)

# Mengubah status Nasi Goreng Jawa
hasil_update2 = koleksi_menu.update_one(
    {"nama_menu": "Nasi Goreng Jawa"},
    {
        "$set": {
            "status": "Habis"
        }
    }
)

print("Jumlah menu yang diperbarui:", 
      hasil_update1.modified_count + hasil_update2.modified_count)

# Menampilkan seluruh menu dan statusnya
for menu in koleksi_menu.find(
    {},
    {"_id": 0, "nama_menu": 1, "status": 1}
):
    print(menu)
```

D. Integrasi ke Spark 
-

```python
# Mengambil Data dari MongoDB

# Mengambil seluruh dokumen dari MongoDB
dokumen_mongo = list(koleksi_menu.find({}))

# Menghitung jumlah dokumen
print("Jumlah dokumen dari MongoDB:", len(dokumen_mongo))
```

```python
# Mengubah Dokumen Menjadi Pandas DataFrame

# Mengubah dokumen MongoDB menjadi Pandas DataFrame
df_pandas = pd.DataFrame(dokumen_mongo)

# Mengubah ObjectId MongoDB menjadi string
df_pandas["_id"] = df_pandas["_id"].astype(str)

# Mengubah nilai kosong (NaN) menjadi None
df_pandas = df_pandas.astype(object).where(
    pd.notnull(df_pandas),
    None
)

# Menampilkan data Pandas
print("Data berhasil diubah menjadi Pandas DataFrame.")
df_pandas.head()
```

```python
# Membuat SparkSession

from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Tugas9-MongoDB-Spark") \
    .master("local[*]") \
    .getOrCreate()

spark.sparkContext.setLogLevel("ERROR")

print("SparkSession berhasil dibuat.")
print("Versi Spark:", spark.version)
```

```python
# Mengonversi Pandas DataFrame menjadi Spark DataFrame

df_spark_menu = spark.createDataFrame(df_pandas)

print("Data berhasil dikonversi menjadi Spark DataFrame.")

# Menampilkan struktur kolom dan tipe data
df_spark_menu.printSchema()

# Menampilkan isi DataFrame
df_spark_menu.show(12, truncate=False)
```

```python
# Menghitung rata-rata harga setiap kategori

from pyspark.sql.functions import avg, round

hasil_rata_rata = df_spark_menu.groupBy(
    "kategori"
).agg(
    round(
        avg("harga"),
        2
    ).alias("rata_rata_harga")
)

# Menampilkan hasil analisis
hasil_rata_rata.show()
```

E. Kesimpulan 
-

Berdasarkan praktikum yang telah dilakukan, salah satu tantangan saat mengubah data MongoDB menjadi Spark DataFrame adalah adanya perbedaan field pada setiap dokumen. Contohnya, menu Kopi memiliki field `tingkat_kepahitan`, sedangkan menu Makanan memiliki field `pedas_level`. Hal tersebut terjadi karena MongoDB menggunakan konsep flexible schema sehingga setiap dokumen dapat memiliki struktur yang berbeda. Sementara itu, Spark DataFrame membutuhkan struktur kolom dan tipe data yang lebih seragam agar data dapat diolah dengan baik. Untuk mengatasi masalah tersebut, data dari MongoDB terlebih dahulu diambil menggunakan PyMongo, kemudian diubah menjadi Pandas DataFrame. Nilai yang kosong disesuaikan menjadi `None` agar dapat dikenali sebagai nilai null oleh Spark. Setelah itu, data dikonversi menjadi Spark DataFrame menggunakan `spark.createDataFrame()`. Dengan cara tersebut, field yang tidak dimiliki oleh suatu dokumen tetap dapat ditampilkan sebagai nilai kosong. Data akhirnya dapat diolah menggunakan Spark, seperti menampilkan skema dan menghitung rata-rata harga berdasarkan kategori. Praktikum ini menunjukkan bahwa MongoDB dan Spark dapat digunakan bersama untuk menyimpan serta menganalisis data dengan struktur yang berbeda.
